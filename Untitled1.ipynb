```python
# Analyze token probabilities for the other known problematic blocks
# using the already-generated 200 DPI result.

def analyze_block_tokens(block_idx):
    generated_ids = result.sequences[block_idx, prompt_len:]

    rows = []

    for step, token_id in enumerate(generated_ids):
        if step >= len(result.scores):
            break

        logits = result.scores[step][block_idx]
        probs = torch.softmax(logits.float(), dim=-1)

        token_prob = probs[token_id].item()

        token_text = processor.decode(
            [token_id.item()],
            skip_special_tokens=False
        )

        rows.append({
            "step": step,
            "token": token_text,
            "probability": token_prob
        })

    return rows


# Known problematic blocks:
# 0 → રોલ નં 16 → રાજા નં. 26
# 1 → કુદરતી → હૃદયની
# 2 → સૂર્યપ્રકાશ → સુપંયપ્રકાશ
# 4 → પાક → વાક
# 4 → નિષ્ફળ → નિરંકુશ
# 5 → અધિકારો → અધિકારી

for block_idx in [0, 1, 4, 5]:

    rows = analyze_block_tokens(block_idx)

    print("\n" + "=" * 90)
    print(f"BLOCK {block_idx}")
    print("=" * 90)

    for row in rows:
        print(
            f"{row['step']:3d} | "
            f"{row['probability']:8.4f} | "
            f"{row['token']!r}"
        )
```

```python
# ============================================================
# STEP 1: FIND INDIVIDUAL TEXT LINES INSIDE ORDER 3
# ============================================================

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Find the request corresponding to ORDER 3
order3_req = None

for req, order in zip(requests_300, orders_300):
    if order == 3:
        order3_req = req
        break

if order3_req is None:
    raise RuntimeError("Could not find ORDER 3 request")

img = order3_req.image.convert("L")
arr = np.array(img)

print("Order 3 image size:", img.size)

# ------------------------------------------------------------
# Detect dark handwritten pixels
# ------------------------------------------------------------
# Adaptive-ish threshold. We only need approximate line
# boundaries here, not OCR.
ink = arr < 210

# Count dark pixels in every horizontal row
row_score = ink.sum(axis=1)

# A row is considered part of handwriting if enough pixels
# are present.
threshold = max(3, int(arr.shape[1] * 0.005))

active = row_score > threshold

# ------------------------------------------------------------
# Convert active rows into continuous bands
# ------------------------------------------------------------
bands = []

start = None

for y, is_active in enumerate(active):
    if is_active and start is None:
        start = y

    elif not is_active and start is not None:
        if y - start >= 3:
            bands.append([start, y])
        start = None

if start is not None:
    bands.append([start, len(active)])

# ------------------------------------------------------------
# Merge bands separated by tiny gaps
# ------------------------------------------------------------
merged = []

for band in bands:
    if not merged:
        merged.append(band)
    else:
        gap = band[0] - merged[-1][1]

        if gap <= 8:
            merged[-1][1] = band[1]
        else:
            merged.append(band)

# ------------------------------------------------------------
# Add small vertical padding
# ------------------------------------------------------------
line_crops = []

for i, (y1, y2) in enumerate(merged):
    pad = 8

    y1p = max(0, y1 - pad)
    y2p = min(arr.shape[0], y2 + pad)

    crop = img.crop((0, y1p, arr.shape[1], y2p))
    line_crops.append(crop)

    print(
        f"Line {i}: "
        f"y={y1p}:{y2p}, "
        f"size={crop.size}"
    )

# ------------------------------------------------------------
# Visualize the detected lines
# ------------------------------------------------------------
plt.figure(figsize=(16, 5))
plt.imshow(img, cmap="gray")

for i, (y1, y2) in enumerate(merged):
    plt.axhline(y1, linestyle="--")
    plt.axhline(y2, linestyle="--")
    plt.text(
        10,
        y1 + 5,
        f"LINE {i}",
        fontsize=12
    )

plt.axis("off")
plt.show()
```

```python
# ============================================================
# STEP 1B: REMOVE NOTEBOOK/EXAM PAPER LINES
# THEN DETECT HANDWRITTEN TEXT LINES
# ============================================================

import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# ------------------------------------------------------------
# Get ORDER 3 image
# ------------------------------------------------------------
order3_req = None

for req, order in zip(requests_300, orders_300):
    if order == 3:
        order3_req = req
        break

img = order3_req.image.convert("L")
arr = np.array(img)

print("Original size:", img.size)

# ------------------------------------------------------------
# Convert to binary ink mask
# ------------------------------------------------------------
# Dark pixels = handwriting + ruled lines
binary = (arr < 190).astype(np.uint8) * 255

# ------------------------------------------------------------
# Detect long horizontal ruled lines
# ------------------------------------------------------------
# These lines are much longer and more continuous than
# individual handwriting strokes.
horizontal_kernel = cv2.getStructuringElement(
    cv2.MORPH_RECT,
    (120, 1)
)

horizontal_lines = cv2.morphologyEx(
    binary,
    cv2.MORPH_OPEN,
    horizontal_kernel
)

# Remove detected horizontal ruling lines
cleaned = cv2.subtract(binary, horizontal_lines)

# ------------------------------------------------------------
# Remove small noise
# ------------------------------------------------------------
small_kernel = cv2.getStructuringElement(
    cv2.MORPH_RECT,
    (2, 2)
)

cleaned = cv2.morphologyEx(
    cleaned,
    cv2.MORPH_OPEN,
    small_kernel
)

# ------------------------------------------------------------
# Calculate horizontal projection of remaining handwriting
# ------------------------------------------------------------
row_score = (cleaned > 0).sum(axis=1)

# Ignore extremely weak rows
threshold = max(2, int(arr.shape[1] * 0.0015))

active = row_score > threshold

# ------------------------------------------------------------
# Find continuous handwriting bands
# ------------------------------------------------------------
bands = []
start = None

for y, active_row in enumerate(active):

    if active_row and start is None:
        start = y

    elif not active_row and start is not None:
        if y - start >= 3:
            bands.append([start, y])
        start = None

if start is not None:
    bands.append([start, len(active)])

# ------------------------------------------------------------
# Merge bands that are very close together
# ------------------------------------------------------------
merged = []

for y1, y2 in bands:

    if not merged:
        merged.append([y1, y2])
        continue

    gap = y1 - merged[-1][1]

    if gap <= 12:
        merged[-1][1] = y2
    else:
        merged.append([y1, y2])

# ------------------------------------------------------------
# Filter unrealistic tiny bands
# ------------------------------------------------------------
final_bands = []

for y1, y2 in merged:

    height = y2 - y1

    if height >= 8:
        final_bands.append([y1, y2])

# ------------------------------------------------------------
# Add padding
# ------------------------------------------------------------
line_crops = []

print("\nDetected handwriting lines:")

for i, (y1, y2) in enumerate(final_bands):

    pad = 10

    y1p = max(0, y1 - pad)
    y2p = min(arr.shape[0], y2 + pad)

    crop = img.crop(
        (0, y1p, arr.shape[1], y2p)
    )

    line_crops.append(crop)

    print(
        f"Line {i}: "
        f"y={y1p}:{y2p}, "
        f"height={y2p-y1p}, "
        f"size={crop.size}"
    )

# ------------------------------------------------------------
# Visualize:
# 1. Original
# 2. Detected ruled lines
# 3. Cleaned handwriting
# 4. Detected text-line boundaries
# ------------------------------------------------------------

fig, axes = plt.subplots(4, 1, figsize=(18, 12))

axes[0].imshow(arr, cmap="gray")
axes[0].set_title("Original Order 3")
axes[0].axis("off")

axes[1].imshow(horizontal_lines, cmap="gray")
axes[1].set_title("Detected Horizontal Ruled Lines")
axes[1].axis("off")

axes[2].imshow(cleaned, cmap="gray")
axes[2].set_title("Handwriting After Ruled-Line Removal")
axes[2].axis("off")

axes[3].imshow(arr, cmap="gray")
axes[3].set_title("Detected Handwriting Line Boundaries")

for i, (y1, y2) in enumerate(final_bands):
    axes[3].axhline(y1, linestyle="--")
    axes[3].axhline(y2, linestyle="--")
    axes[3].text(
        10,
        y1 + 5,
        f"LINE {i}",
        fontsize=12
    )

axes[3].axis("off")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# STEP 1C: DETECT RULED PAPER LINES
# THEN USE THE SPACES BETWEEN THEM AS TEXT-LINE REGIONS
# ============================================================

import cv2
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# Get ORDER 3 image
# ------------------------------------------------------------
order3_req = None

for req, order in zip(requests_300, orders_300):
    if order == 3:
        order3_req = req
        break

img = order3_req.image.convert("L")
arr = np.array(img)

h, w = arr.shape

print("Image size:", (w, h))

# ------------------------------------------------------------
# Threshold dark pixels
# ------------------------------------------------------------
# Use a relatively strict threshold so faint paper texture
# is ignored.
binary = cv2.threshold(
    arr,
    170,
    255,
    cv2.THRESH_BINARY_INV
)[1]

# ------------------------------------------------------------
# Detect long horizontal structures
# ------------------------------------------------------------
# Large horizontal kernel isolates the long ruled-paper lines.
kernel_width = max(300, w // 5)

horizontal_kernel = cv2.getStructuringElement(
    cv2.MORPH_RECT,
    (kernel_width, 1)
)

horizontal = cv2.morphologyEx(
    binary,
    cv2.MORPH_OPEN,
    horizontal_kernel
)

# ------------------------------------------------------------
# Get horizontal-line strength for each row
# ------------------------------------------------------------
horizontal_score = (horizontal > 0).sum(axis=1)

# A real ruled line should extend over a substantial
# portion of the block.
line_threshold = int(w * 0.20)

candidate_rows = np.where(
    horizontal_score > line_threshold
)[0]

# ------------------------------------------------------------
# Group adjacent candidate rows
# ------------------------------------------------------------
line_groups = []

if len(candidate_rows) > 0:

    start = candidate_rows[0]
    prev = candidate_rows[0]

    for y in candidate_rows[1:]:

        if y <= prev + 2:
            prev = y

        else:
            line_groups.append((start, prev))
            start = y
            prev = y

    line_groups.append((start, prev))

# ------------------------------------------------------------
# Convert each group into a single y coordinate
# ------------------------------------------------------------
rule_y = []

for y1, y2 in line_groups:
    rule_y.append((y1 + y2) // 2)

# Remove very-close duplicates
clean_rule_y = []

for y in rule_y:

    if not clean_rule_y or y - clean_rule_y[-1] > 5:
        clean_rule_y.append(y)

print("\nDetected ruled lines:")
print(clean_rule_y)

# ------------------------------------------------------------
# Create text regions BETWEEN ruled lines
# ------------------------------------------------------------
# Add artificial boundaries at top/bottom.
boundaries = [0] + clean_rule_y + [h]

regions = []

for i in range(len(boundaries) - 1):

    top = boundaries[i]
    bottom = boundaries[i + 1]

    # Ignore extremely small regions
    if bottom - top < 20:
        continue

    # Keep a little overlap around each ruling.
    pad = 8

    y1 = max(0, top - pad)
    y2 = min(h, bottom + pad)

    regions.append((y1, y2))

print("\nCandidate text regions:")

for i, (y1, y2) in enumerate(regions):
    print(
        f"Region {i}: "
        f"y={y1}:{y2}, "
        f"height={y2-y1}"
    )

# ------------------------------------------------------------
# Visualization
# ------------------------------------------------------------
fig, axes = plt.subplots(
    3,
    1,
    figsize=(18, 12)
)

# Original
axes[0].imshow(arr, cmap="gray")
axes[0].set_title("Original Order 3")
axes[0].axis("off")

# Detected ruled lines
axes[1].imshow(arr, cmap="gray")
axes[1].set_title("Detected Ruled Lines")

for i, y in enumerate(clean_rule_y):
    axes[1].axhline(
        y,
        linestyle="--"
    )
    axes[1].text(
        10,
        y - 3,
        f"RULE {i}",
        fontsize=10
    )

axes[1].axis("off")

# Candidate text regions
axes[2].imshow(arr, cmap="gray")
axes[2].set_title("Candidate Text Regions")

for i, (y1, y2) in enumerate(regions):

    axes[2].axhline(
        y1,
        linestyle="--"
    )

    axes[2].axhline(
        y2,
        linestyle="--"
    )

    axes[2].text(
        10,
        y1 + 5,
        f"REGION {i}",
        fontsize=10
    )

axes[2].axis("off")

plt.tight_layout()
plt.show()
```

```python
# ============================================================
# STEP 4: VISUAL WORD-REGION DETECTION
# ============================================================

import cv2
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# line_img already exists from the previous step
img = line_img.convert("L")
arr = np.array(img)

h, w = arr.shape

print("Line size:", (w, h))

# ------------------------------------------------------------
# 1. Threshold handwriting
# ------------------------------------------------------------

binary = cv2.threshold(
    arr,
    180,
    255,
    cv2.THRESH_BINARY_INV
)[1]

# ------------------------------------------------------------
# 2. Connect characters belonging to the same handwritten
#    word while keeping large word gaps.
# ------------------------------------------------------------

# Horizontal dilation connects nearby Gujarati characters.
# We intentionally keep the kernel relatively small.
kernel = cv2.getStructuringElement(
    cv2.MORPH_RECT,
    (18, 3)
)

connected = cv2.morphologyEx(
    binary,
    cv2.MORPH_CLOSE,
    kernel
)

# A little dilation horizontally
horizontal_kernel = cv2.getStructuringElement(
    cv2.MORPH_RECT,
    (12, 1)
)

connected = cv2.dilate(
    connected,
    horizontal_kernel,
    iterations=1
)

# ------------------------------------------------------------
# 3. Find connected candidate regions
# ------------------------------------------------------------

num_labels, labels, stats, centroids = cv2.connectedComponentsWithStats(
    connected,
    connectivity=8
)

candidates = []

for i in range(1, num_labels):

    x, y, bw, bh, area = stats[i]

    # Ignore tiny noise
    if area < 40:
        continue

    # Ignore extremely tiny vertical artifacts
    if bw < 10 or bh < 8:
        continue

    candidates.append(
        (x, y, bw, bh, area)
    )

# Sort left-to-right
candidates.sort(key=lambda z: z[0])

print("\nCandidate regions:")

for i, (x, y, bw, bh, area) in enumerate(candidates):
    print(
        f"{i}: "
        f"x={x}:{x+bw}, "
        f"y={y}:{y+bh}, "
        f"size={bw}x{bh}, "
        f"area={area}"
    )

# ------------------------------------------------------------
# 4. Visualize candidates on original line
# ------------------------------------------------------------

rgb = np.array(line_img.convert("RGB"))

plt.figure(figsize=(22, 5))
plt.imshow(rgb)

for i, (x, y, bw, bh, area) in enumerate(candidates):

    rect = plt.Rectangle(
        (x, y),
        bw,
        bh,
        fill=False,
        linewidth=2
    )

    plt.gca().add_patch(rect)

    plt.text(
        x,
        max(0, y - 3),
        str(i),
        fontsize=13
    )

plt.title("Candidate Word Regions")
plt.axis("off")
plt.show()
```

```python
# ============================================================
# STEP 6: TOKEN UNCERTAINTY FOR ORDER 3
# ============================================================

import torch
import numpy as np

# Use the original 300 DPI ORDER 3 request
order3_request = None

for req, order in zip(requests_300, orders_300):
    if order == 3:
        order3_request = req
        break

# ------------------------------------------------------------
# Prepare exactly like backend.transcribe()
# ------------------------------------------------------------

backend = parser.recognizer.backend
model = backend.model
processor = backend.processor

inputs = processor(
    text=[order3_request.prompt],
    images=[order3_request.image],
    padding=True,
    return_tensors="pt"
)

inputs = {
    k: v.to(model.device) if hasattr(v, "to") else v
    for k, v in inputs.items()
}

# ------------------------------------------------------------
# Generate with token scores
# ------------------------------------------------------------

with torch.inference_mode():

    output = model.generate(
        **inputs,
        max_new_tokens=2048,
        do_sample=False,
        use_cache=True,
        return_dict_in_generate=True,
        output_scores=True,
        eos_token_id=backend.eos_token_id,
        pad_token_id=backend.pad_token_id,
    )

# ------------------------------------------------------------
# Calculate probability of selected generated token
# ------------------------------------------------------------

prompt_len = inputs["input_ids"].shape[1]

generated_ids = output.sequences[0, prompt_len:]

scores = output.scores

token_info = []

for step, score in enumerate(scores):

    token_id = generated_ids[step].item()

    # Stop after EOS
    if token_id == backend.eos_token_id:
        break

    probs = torch.softmax(score[0], dim=-1)
    prob = probs[token_id].item()

    token_text = processor.tokenizer.decode(
        [token_id],
        skip_special_tokens=False
    )

    token_info.append(
        {
            "step": step,
            "token": token_text,
            "probability": prob
        }
    )

# ------------------------------------------------------------
# Print tokens with uncertainty
# ------------------------------------------------------------

print("ORDER 3 OCR:")
print(
    processor.tokenizer.decode(
        generated_ids,
        skip_special_tokens=False
    )
)

print("\nTOKEN UNCERTAINTY:")
print("=" * 70)

for item in token_info:

    marker = ""

    if item["probability"] < 0.30:
        marker = "  <<< VERY UNCERTAIN"
    elif item["probability"] < 0.50:
        marker = "  <<< UNCERTAIN"

    print(
        f"{item['step']:3d} | "
        f"{item['probability']:.3f} | "
        f"{repr(item['token'])}"
        f"{marker}"
    )
```

```python
# ============================================================
# STEP 6: ORDER 3 TOKEN UNCERTAINTY
# USING THE EXACT WORKING BACKEND PIPELINE
# ============================================================

import torch

backend = parser.recognizer.backend
model = backend.model
processor = backend.processor

# ------------------------------------------------------------
# Get ORDER 3 request
# ------------------------------------------------------------

order3_request = None

for req, order in zip(requests_300, orders_300):
    if order == 3:
        order3_request = req
        break

if order3_request is None:
    raise RuntimeError("ORDER 3 request not found")

# ------------------------------------------------------------
# EXACT SAME preprocessing used by backend.transcribe()
# ------------------------------------------------------------

inputs = processor(
    text=[backend._prompt(order3_request.prompt)],
    images=[order3_request.image],
    padding=True,
    return_tensors="pt",
).to(model.device)

prompt_len = inputs["input_ids"].shape[-1]

print("Prompt tokens:", prompt_len)
print("Input keys:", list(inputs.keys()))

# ------------------------------------------------------------
# Generate with scores
# ------------------------------------------------------------

with backend._torch.inference_mode():

    output = model.generate(
        **inputs,
        max_new_tokens=backend.config.max_tokens,
        do_sample=False,
        use_cache=True,
        eos_token_id=backend.eos_token_id,
        pad_token_id=backend.pad_token_id,
        return_dict_in_generate=True,
        output_scores=True,
    )

# ------------------------------------------------------------
# Generated tokens
# ------------------------------------------------------------

generated_ids = output.sequences[0, prompt_len:]

decoded = processor.batch_decode(
    output.sequences[:, prompt_len:],
    skip_special_tokens=True
)[0].strip()

print("\n========================================")
print("ORDER 3 OCR")
print("========================================")
print(decoded)

# ------------------------------------------------------------
# Token probabilities
# ------------------------------------------------------------

token_info = []

for step, score in enumerate(output.scores):

    if step >= len(generated_ids):
        break

    token_id = generated_ids[step].item()

    # Stop at EOS
    if token_id == backend.eos_token_id:
        break

    probability = torch.softmax(
        score[0],
        dim=-1
    )[token_id].item()

    token_text = processor.tokenizer.decode(
        [token_id],
        skip_special_tokens=False
    )

    token_info.append(
        {
            "step": step,
            "token": token_text,
            "probability": probability,
        }
    )

# ------------------------------------------------------------
# Display token probabilities
# ------------------------------------------------------------

print("\n========================================")
print("TOKEN UNCERTAINTY")
print("========================================")

for item in token_info:

    p = item["probability"]

    if p < 0.30:
        marker = " <<< VERY LOW"
    elif p < 0.50:
        marker = " <<< LOW"
    else:
        marker = ""

    print(
        f"{item['step']:3d} | "
        f"{p:6.3f} | "
        f"{repr(item['token'])}"
        f"{marker}"
    )

# ------------------------------------------------------------
# Summary statistics
# ------------------------------------------------------------

probs = [
    x["probability"]
    for x in token_info
]

if probs:
    print("\n========================================")
    print("SUMMARY")
    print("========================================")
    print(f"Generated tokens: {len(probs)}")
    print(f"Mean probability: {sum(probs)/len(probs):.3f}")
    print(f"Min probability:  {min(probs):.3f}")
    print(
        f"Tokens < 0.50:   "
        f"{sum(p < 0.50 for p in probs)}"
    )
    print(
        f"Tokens < 0.30:   "
        f"{sum(p < 0.30 for p in probs)}"
    )
```

```python
# ============================================================
# PAGE 1: GROUND-TRUTH EVALUATION
# ============================================================

import re

# ------------------------------------------------------------
# Manually verified visual ground truth
# ------------------------------------------------------------

ground_truth = [
    "રોલ નં 16",
    "કુદરતી",
    "સૂર્યપ્રકાશ",
    "પાક",
    "નિષ્ફળ",
    "અધિકારો",
]

# ------------------------------------------------------------
# Corresponding 1200px OCR outputs
#
# These are extracted from the OCR output we just generated.
# ------------------------------------------------------------

ocr_output = [
    "રાત્રિ નં. 26",
    "ઉદરતી",
    "સૂર્યપ્રકાશ",
    "પાક",
    "નિરફળ",
    "અધિકારો",
]

# ------------------------------------------------------------
# Gujarati-safe normalization
#
# IMPORTANT:
# This does NOT correct spelling.
# It only removes whitespace differences so that:
#
# "રોલ નં 16"
# and
# "રોલ  નં 16"
#
# aren't treated as different.
# ------------------------------------------------------------

def normalize_for_comparison(text):
    text = text.strip()
    text = re.sub(r"\s+", " ", text)
    return text


# ------------------------------------------------------------
# Character-level Levenshtein distance
# ------------------------------------------------------------

def levenshtein(a, b):

    a = list(a)
    b = list(b)

    previous = list(range(len(b) + 1))

    for i, ca in enumerate(a, start=1):

        current = [i]

        for j, cb in enumerate(b, start=1):

            insert = current[j - 1] + 1
            delete = previous[j] + 1
            substitute = previous[j - 1] + (ca != cb)

            current.append(
                min(insert, delete, substitute)
            )

        previous = current

    return previous[-1]


# ------------------------------------------------------------
# Calculate results
# ------------------------------------------------------------

print("========================================")
print("PAGE 1 OCR EVALUATION")
print("========================================")

total_chars = 0
total_errors = 0
exact_matches = 0

for gt, pred in zip(ground_truth, ocr_output):

    gt_norm = normalize_for_comparison(gt)
    pred_norm = normalize_for_comparison(pred)

    distance = levenshtein(
        gt_norm,
        pred_norm
    )

    max_len = max(len(gt_norm), 1)

    cer = distance / max_len

    total_chars += len(gt_norm)
    total_errors += distance

    if gt_norm == pred_norm:
        exact_matches += 1

    print("\nGROUND TRUTH:")
    print(gt_norm)

    print("OCR:")
    print(pred_norm)

    print(f"Edit distance: {distance}")
    print(f"CER: {cer:.2%}")

    if gt_norm == pred_norm:
        print("EXACT MATCH: YES")
    else:
        print("EXACT MATCH: NO")


# ------------------------------------------------------------
# Overall metrics
# ------------------------------------------------------------

overall_cer = total_errors / max(total_chars, 1)
exact_rate = exact_matches / len(ground_truth)

print("\n========================================")
print("OVERALL")
print("========================================")

print(f"Total examples: {len(ground_truth)}")
print(f"Total ground-truth characters: {total_chars}")
print(f"Total character errors: {total_errors}")
print(f"Overall CER: {overall_cer:.2%}")
print(f"Exact match rate: {exact_rate:.2%}")
```

```python
import pymupdf
import os
import time

pdf_path = "/content/Standard_8_Hindi_40_Marks_Handwritten_Answer_Key (1).pdf"
output_dir = "/content/hindi_pages"

os.makedirs(output_dir, exist_ok=True)

doc = pymupdf.open(pdf_path)

results = {}

for page_num in range(2, 5):

    print("\n" + "=" * 80)
    print(f"PROCESSING HINDI PAGE {page_num}")
    print("=" * 80)

    # Convert page to 200 DPI
    page = doc[page_num - 1]

    dpi = 200
    scale = dpi / 72
    matrix = pymupdf.Matrix(scale, scale)

    pix = page.get_pixmap(matrix=matrix, alpha=False)

    image_path = f"{output_dir}/page_{page_num}.png"
    pix.save(image_path)

    print(
        "Image:",
        pix.width,
        "x",
        pix.height
    )

    # OCR
    start = time.time()

    result = parser.parse(image_path)

    elapsed = time.time() - start

    results[page_num] = {
        "result": result,
        "time": elapsed
    }

    print("\nRAW OCR:")
    print(result)

    print("\nTime:", round(elapsed, 2), "seconds")

print("\n" + "=" * 80)
print("SUMMARY")
print("=" * 80)

for page_num, data in results.items():
    print(
        f"Page {page_num}: "
        f"{data['time']:.2f} sec"
    )
```
